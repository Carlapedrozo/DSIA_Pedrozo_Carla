```
def calcular_suma(a,b):
    suma=a+b
    return suma
print(calcular_suma(a,b))
```

```
def calcular_resta(a,b):
    resta=a-b
    return resta
print(calcular_resta(a,b))
```

```
def calcular_division(a,b):
    division=a/b
    return division
print(calcular_division(a,b))
```

```
def calcular_multiplicacion(a,b):
    multiplicacion=a*b
    return multiplicacion
print(calcular_multiplicacion(a,b))
```
