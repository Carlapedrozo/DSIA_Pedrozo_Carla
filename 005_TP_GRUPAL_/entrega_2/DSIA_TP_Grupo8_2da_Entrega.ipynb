### Análisis Estadístico Descriptivo

Se utilizarán funciones de estadística descriptiva para obtener un resumen numérico y categórico del conjunto de datos, permitiendo observar medias, desviaciones, cuartiles y frecuencias absolutas\.

```

# CORRECCIÓN 3 - Identificación de valores atípicos (IQR)

variables = ['distancia_km', 'tiempo_entrega_minutos']

for variable in variables:

    Q1 = df_delivery[variable].quantile(0.25)
    Q3 = df_delivery[variable].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    atipicos = df_delivery[
        (df_delivery[variable] < limite_inferior) |
        (df_delivery[variable] > limite_superior)
    ]

    print(f"\nVariable: {variable}")
    print(f"Límite inferior: {limite_inferior:.2f}")
    print(f"Límite superior: {limite_superior:.2f}")
    print(f"Cantidad de valores atípicos: {len(atipicos)}")

# Inspeccionamos el pedido con mayor distancia registrada
print("\nPedido con mayor distancia:")
display(
    df_delivery.nlargest(1, 'distancia_km')[
        ['id_pedido', 'distancia_km',
         'tiempo_entrega_minutos', 'estado_pedido']
    ]
)
```

**Interpretación de los Valores Atípicos**

El análisis descriptivo permitió identificar valores extremos en las variables `distancia_km` y `tiempo_entrega_minutos`. Mediante el método del rango intercuartílico (IQR), se establecieron límites superiores de 6,25 km y 58 minutos, respectivamente, detectándose 179 observaciones atípicas en distancia y 73 en tiempo de entrega.

Estos resultados evidencian la presencia de registros que se alejan de la distribución central. Sin embargo, su condición de valores atípicos no implica necesariamente que sean erróneos. Por lo tanto, se conservarán durante esta etapa exploratoria y su tratamiento se evaluará posteriormente según su coherencia y relevancia para el análisis.

```
print("--- Estadísticas de Variables Numéricas ---")
display(df_delivery.describe().round(2))

print("\n--- Estadísticas de Variables Categóricas ---")
display(df_delivery.describe(include=['object', 'category']))
```

**Interpretación de las Variables**

Las estadísticas descriptivas muestran diferencias entre los valores centrales y máximos de las variables numéricas, particularmente en la distancia y el tiempo de entrega, cuyos máximos alcanzan 59,3 km y 180 minutos, respectivamente\. ## 2\. Visualizaciones y análisis

### Gráfico 1: Cantidad de pedidos por App \(Countplot\)

Este gráfico nos mostrará cuántos pedidos se hicieron en cada aplicación de delivery\.

```

# CORRECCIÓN 1 - Gráfico de pedidos por aplicación

# Copia para conservar los valores originales
df_apps = df_delivery.copy()

# Unificación de las denominaciones
mapeo_apps = {
    'pedidos ya': 'PedidosYa',
    'Pedidos Ya': 'PedidosYa',
    'PeYa': 'PedidosYa',
    'RAPPI': 'Rappi',
    'rappi': 'Rappi',
    'rapi': 'Rappi'
}

df_apps['app_delivery'] = (
    df_apps['app_delivery'].replace(mapeo_apps)
)

# Conteo y porcentaje por aplicación
conteo_apps = df_apps['app_delivery'].value_counts()
porcentajes = (conteo_apps / len(df_apps) * 100).round(2)

resumen_apps = pd.DataFrame({
    'Cantidad de pedidos': conteo_apps,
    'Porcentaje (%)': porcentajes
})

print("Distribución de pedidos por aplicación:")
display(resumen_apps)

# Participación conjunta
total_principales = conteo_apps.loc[['PedidosYa', 'Rappi']].sum()
porcentaje_principales = total_principales / len(df_apps) * 100

print(f"\nPedidosYa + Rappi: {porcentaje_principales:.2f}%")

# Gráfico corregido
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df_apps,
    x='app_delivery',
    order=conteo_apps.index,
    color='steelblue'
)

plt.title("Cantidad de pedidos por aplicación - Datos simulados")
plt.xlabel("Aplicación de delivery")
plt.ylabel("Cantidad de pedidos")
plt.tight_layout()
plt.show()
```

### 📌 Interpretación de Gráfico 1

Luego de estandarizar las denominaciones de las aplicaciones, se identificó que PedidosYa concentra 1\.564 pedidos \(44,69 %\), seguida de Rappi con 1\.413 \(40,37 %\)\. Ambas plataformas representan conjuntamente el 85,06 % de los registros analizados\.

Por su parte, UberEats y Zonakids presentan participaciones menores, con 9,69 % y 5,26 %, respectivamente\. Estos resultados evidencian una concentración de pedidos en dos aplicaciones dentro del dataset simulado\.

### Gráfico 2: Relación entre la distancia y el monto del pedido \(Scatterplot\)

Se usa un diagrama de dispersión para ver si existe alguna relación entre los kilómetros de distancia del envío y el monto total del pedido\. Además, se colorean los puntos según la app

```

# CORRECCIÓN 2 - Relación entre distancia y monto del pedido

# Calculamos la correlación de Pearson
correlacion = df_apps['distancia_km'].corr(
    df_apps['monto_pedido'],
    method='pearson'
)

print(f"Correlación de Pearson: {correlacion:.4f}")

# Gráfico de dispersión
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_apps,
    x='distancia_km',
    y='monto_pedido',
    hue='app_delivery',
    alpha=0.5,
    s=35
)

plt.title("Relación entre distancia y monto del pedido")
plt.xlabel("Distancia del envío (km)")
plt.ylabel("Monto del pedido (ARS)")
plt.legend(title="Aplicación")
plt.tight_layout()
plt.show()
```

### 📌 Interpretación de Gráfico 2

El gráfico de dispersión no evidencia una tendencia lineal definida entre la distancia del envío y el monto del pedido\. Esta observación se respalda con un coeficiente de correlación de Pearson de 0,0007, lo que indica una asociación lineal prácticamente nula entre ambas variables\.

Los resultados sugieren que, dentro del dataset analizado, la distancia del envío no presenta una relación lineal relevante con el importe de los pedidos\. Asimismo, se observan algunas distancias considerablemente superiores al resto, que serán evaluadas posteriormente para determinar si corresponden a valores atípicos válidos o posibles errores\.

### Gráfico 2: Relación entre presupuesto de desarrollo y ventas

Se analizará la relación entre el presupuesto de desarrollo y las ventas promedio mensuales de los videojuegos mediante un gráfico de dispersión, diferenciando los títulos por género\. Para ello, se calculará el promedio de unidades vendidas por mes de cada videojuego, considerando que algunos presentan períodos de registro incompletos\.

Además, se utilizará el coeficiente de correlación de Pearson para evaluar la dirección e intensidad de la relación lineal entre ambas variables\.

```
# CORRECCIÓN VIDEOJUEGOS - GRÁFICO 2
# Relación entre presupuesto y ventas promedio mensual

# Agrupamos los registros por videojuego
df_ventas_juego = (
    df_videojuegos.groupby(
        ['id_videojuego', 'titulo', 'genero_principal_referencial'],
        as_index=False
    ).agg(
        presupuesto_desarrollo=(
            'presupuesto_desarrollo_simulado_usd', 'first'
        ),
        ventas_totales=(
            'unidades_mes_simuladas', 'sum'
        ),
        ventas_promedio_mensual=(
            'unidades_mes_simuladas', 'mean'
        ),
        meses_registrados=(
            'mes_venta_simulado', 'nunique'
        )
    )
)

# Calculamos las correlaciones
corr_total = df_ventas_juego['presupuesto_desarrollo'].corr(
    df_ventas_juego['ventas_totales']
)

corr_promedio = df_ventas_juego['presupuesto_desarrollo'].corr(
    df_ventas_juego['ventas_promedio_mensual']
)

print(f"Correlación con ventas acumuladas: {corr_total:.4f}")
print(f"Correlación con promedio mensual: {corr_promedio:.4f}")

# Convertimos escalas para facilitar la visualización
df_ventas_juego['presupuesto_millones'] = (
    df_ventas_juego['presupuesto_desarrollo'] / 1_000_000
)

df_ventas_juego['ventas_promedio_miles'] = (
    df_ventas_juego['ventas_promedio_mensual'] / 1_000
)

# Visualización
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_ventas_juego,
    x='presupuesto_millones',
    y='ventas_promedio_miles',
    hue='genero_principal_referencial',
    alpha=0.7,
    s=60
)

plt.title("Presupuesto de desarrollo y ventas promedio mensual")
plt.xlabel("Presupuesto de desarrollo (millones USD)")
plt.ylabel("Unidades vendidas por mes (miles)")
plt.legend(title="Género", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

### Interpretación del Gráfico

El gráfico de dispersión muestra una asociación lineal positiva débil entre el presupuesto de desarrollo y las ventas promedio mensuales de los videojuegos, respaldada por un coeficiente de correlación de Pearson de 0,2748\. Asimismo, la correlación entre el presupuesto y las ventas acumuladas fue de 0,2775, mostrando una tendencia similar\.

Se observa una dispersión considerable de los datos, con algunos títulos que presentan ventas elevadas respecto del conjunto\. Esto sugiere que el presupuesto de desarrollo, por sí solo, no permite explicar adecuadamente las diferencias de ventas entre videojuegos\.

Para reducir el efecto de las diferencias en la cobertura temporal, se utilizó el promedio mensual de ventas de cada título\. Sin embargo, los resultados deben interpretarse considerando que cuatro videojuegos presentan períodos incompletos\.

### 2\.7 Revisión de Valores Atípicos

### Evaluación de valores atípicos mediante IQR

```
variables_evaluadas = [
    'distancia_km',
    'tiempo_entrega_minutos',
    'monto_pedido'
]

resultados_atipicos = []

for variable in variables_evaluadas:
    q1 = df_delivery[variable].quantile(0.25)
    q3 = df_delivery[variable].quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    cantidad = (
        (df_delivery[variable] < limite_inferior) |
        (df_delivery[variable] > limite_superior)
    ).sum()

    resultados_atipicos.append({
        'Variable': variable,
        'Límite superior': round(limite_superior, 2),
        'Valores atípicos': cantidad
    })

display(pd.DataFrame(resultados_atipicos))
```

**Interpretación de la Revisión**
La evaluación mediante el rango intercuartílico (IQR) identificó 179 valores atípicos en `distancia_km`, 73 en `tiempo_entrega_minutos` y 28 en `monto_pedido`. Estos resultados muestran la presencia de observaciones alejadas de los rangos habituales de cada variable, sin que ello implique necesariamente errores de registro.

Se decidió conservar los valores atípicos y los registros originales, ya que no se cuenta con evidencia suficiente para justificar su eliminación o modificación. Su posible influencia sobre los resultados estadísticos y el desempeño de futuros modelos predictivos deberá evaluarse en etapas posteriores.

### Hipótesis 2 : Asociación entre la aplicación de delivery y el uso de promociones

Se plantea que la proporción de pedidos con promoción varía según la aplicación de delivery utilizada.

Para evaluar esta hipótesis, se analizarán las variables categóricas `app_delivery` y `aplica_promocion`. Se tomará como referencia el gráfico de barras apiladas del análisis bivariado y se aplicará una prueba de chi-cuadrado de independencia para examinar si existe evidencia estadística de asociación entre ambas variables.

**Visualización de referencia:** Gráfico 3 del Análisis Bivariado — *Uso de promociones según aplicación de delivery*. Esta visualización permite comparar las proporciones de pedidos con y sin promoción por plataforma y complementa la prueba de chi-cuadrado presentada en esta hipótesis.

```
# Hipótesis 2: asociación entre aplicación y promociones

from scipy.stats import chi2_contingency

# Frecuencias absolutas observadas
tabla_contingencia = pd.crosstab(
    df_delivery['app_delivery'],
    df_delivery['aplica_promocion']
)

print("Tabla de contingencia:")
display(tabla_contingencia)

```

Prueba de chi\-cuadrado de independencia

```

chi2, p_valor, grados_libertad, esperados = chi2_contingency(
    tabla_contingencia
)

print(f"Chi-cuadrado: {chi2:.4f}")
print(f"Valor p: {p_valor:.4f}")
print(f"Grados de libertad: {grados_libertad}")
print(f"Frecuencia esperada mínima: {esperados.min():.2f}")
```

**Explicación del Desarrollo**
Para evaluar la hipótesis, se analizaron las variables categóricas `app_delivery` y `aplica_promocion`. Inicialmente, se compararon los porcentajes de pedidos con promoción por aplicación mediante un gráfico de barras apiladas, observándose valores entre el 34,78 % de Zonakids y el 42,39 % de Rappi.

Posteriormente, se construyó una tabla de contingencia con las frecuencias absolutas y se aplicó una prueba de chi-cuadrado de independencia, obteniendo un estadístico χ² = 5,3049, con 3 grados de libertad y un valor p = 0,1508.

**Conclusión de la Hipótesis**
La hipótesis no obtuvo respaldo estadístico suficiente. Aunque se identificaron diferencias descriptivas en las proporciones de pedidos con promoción entre las aplicaciones, el valor p obtenido (0,1508) es superior al nivel de significación establecido (0,05).

Por lo tanto, no se rechaza la hipótesis nula de independencia entre ambas variables. Esto no demuestra que sean independientes, sino que los datos analizados no aportan evidencia estadística suficiente para afirmar una asociación. Además, los resultados corresponden a un dataset sintético y no pueden generalizarse al mercado real.
